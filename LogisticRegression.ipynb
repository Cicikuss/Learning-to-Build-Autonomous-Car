```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
def draw(x1,x2):
  ln = plt.plot(x1,x2)
def sigmoid(score):
  return 1/(1+np.exp(-score))
def calculate_error(linear_parameters,points,y):
  m=points.shape[0]
  p = sigmoid(points*linear_parameters)
  cross_entropy =-(1/m)*(np.log(p).T * y + np.log(1-p).T * (1-y))
  return cross_entropy
def gradient_descent(linear_parameters,points,y,alpha):
  m=points.shape[0]
  for i in range(2500):
    p = sigmoid(points*linear_parameters)
    gradient = (points.T * (p - y))*(alpha/m)
    linear_parameters = linear_parameters -  gradient
    w1 = linear_parameters.item(0)
    w2 = linear_parameters.item(1)
    b = linear_parameters.item(2)
    x1 = np.array([points[:,0].min(),points[:,0].max()])
    x2 = -b/w2 + (x1*(-w1/w2))
    print(calculate_error(linear_parameters,points,y))
  draw(x1,x2)
```

```python
n_pts = 100
np.random.seed(0)
bias = np.ones(n_pts)
top_region = np.array([np.random.normal(10,2,n_pts),np.random.normal(12,2,n_pts),bias  ]).T
bottom_region = np.array([np.random.normal(5,2,n_pts),np.random.normal(6,2,n_pts),bias  ]).T
all_points = np.vstack((top_region,bottom_region))
linear_parameters = np.matrix([np.zeros(3)]).T
#x1 = np.array([bottom_region[:,0].min(),top_region[:,0].max()])
#x2 = -b/w2 + (x1*(-w1/w2))
linear_combination = all_points * linear_parameters
probilities = sigmoid(linear_combination)
y= np.array([np.zeros(n_pts),np.ones(n_pts)]).reshape(n_pts*2,1)
```

```python
fig , ax = plt.subplots(figsize=(4,4))
ax.scatter(top_region[:,0],top_region[:,1],color='r')
ax.scatter(bottom_region[:,0],bottom_region[:,1],color='b')
gradient_descent(linear_parameters,all_points,y,0.06)
plt.show()
```
